```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_chroma import Chroma
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_huggingface import HuggingFaceEmbeddings
import gradio as gr
```

```python
MODEL = "gpt-4.1-nano"
DB_NAME = "vector_db"
load_dotenv(override=True)
```

```python
from langchain_openai import OpenAIEmbeddings


embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
vectorStore = Chroma(persist_directory=DB_NAME, embedding_function=embeddings)
```

```python
reciever = vectorStore.as_retriever()
llm = ChatOpenAI(temperature=0.0,model_name=MODEL)
```

```python
SYSTEM_PROMPT="""
you are a friendly Assisstant representing company called insurellm.
you are chatting with a user about insurellm,
if you find useful information, please give the answer.
Context:{context}

"""
```

```python
def answer_question(question:str,history):
    docs = reciever.invoke(question)
    context = "\n\n".join(doc.page_content for doc in docs)
    system_query = SYSTEM_PROMPT.format(context=context)
    response = llm.invoke([SystemMessage(content=system_query),HumanMessage(content=question)])
    return response.content
```
