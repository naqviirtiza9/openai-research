```python
import os
```

```python
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
if api_key:
    print("successful")
```

```python
message = "Hello Gpt, how are you are you fine"
messages = [
    {"role" : "user","content":message}
]
messages
```

```python
openai = OpenAI()
response = openai.chat.completions.create(model="gpt-5-nano",messages=messages)

response.choices[0].message.content
```

```python
import requests
from bs4 import BeautifulSoup
url = "https://ian-portfolio-roan.vercel.app/"
```

```python
response = requests.get(url)
if response.status_code==200:
    soup = BeautifulSoup(response.text, 'html.parser')
    print("Title is ", soup("title"))
```

```python
soup.find("title")
```

```python
system_prompt = """
You ara an assistant that analyzes the contents of a website,
and provides a short summary, ignoring text that might e navigation related.
"""
```

```python
user_prompt = """
Here are some contents of a website.
provide a brief summary of this in a paragraph
"""
```

```python
messages=[
    {"role":"system","content":"You are a assisstant that explains with politeness and care"}
    ,
    {"role":"user","content":"give me a code snippet of webscrapping"}
]
response = openai.chat.completions.create(model="gpt-4.1-nano",messages=messages)
```

```python
response.choices[0].message.content
```

```python
messages = [
    {"role":"system","content":system_prompt},
    {"role":"user","content":user_prompt + soup.text}
]
response = openai.chat.completions.create(model="gpt-4.1-nano",messages=messages)
```

```python
Markdown(response.choices[0].message.content)
```

```python
##more tasks
header = {"Authorization" : f"Bearer {api_key}" , "Content-type": "application/json"}

payload = {
    "model" : "gpt-5-nano",
    "messages":[
        {"role":"user" , "content":"Tell me who was Ayub khan"}
    ]
}
```

```python
response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=header,
    json=payload
)
```

# **Using different LLMS using the OpenAI Client library**

```python
google_api_key = os.getenv('GOOGLE_API_KEY')
gemini_api_end = "https://generativelanguage.googleapis.com/v1beta"
if not google_api_key:
    print("No api key")
```

```python
gemini_chat_response = OpenAI(base_url=gemini_api_end,api_key=google_api_key)
```

```python
messages=[
    {"role":"system","content":"You are a assisstant that explains with politeness and care"}
    ,
    {"role":"user","content":"what is evaporation"}
]
response = gemini_chat_response.chat.completions.create(model="gemini-2.5-pro",messages=messages)
```

```python
response2 = openai.chat.completions.create(model="gpt-4.1-nano",messages=messages)
response2.choices[0].message.content
```
